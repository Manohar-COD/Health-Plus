```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
```

```python
df = pd.read_csv("patient_data.csv")
print("Original dataset")
print(df)
```

```python
print("Missing Values")
print(df.isnull().sum())

print("\nDuplicate Values")
print(df.duplicated().sum())

print("\nDataset Information")
print(df.info())
```

```python
print("Data types")
print(df.dtypes)
```

```python
df = df.drop_duplicates()
print(df)
```

```python
df["Age"] = df["Age"].fillna(df["Age"].mean())
df["BloodPressure"] = df["BloodPressure"].fillna(df["BloodPressure"].mean())
df["Glucose"] = df["Glucose"].fillna(df["Glucose"].mean())
print(df)
```

```python
scaler = MinMaxScaler()
df[["Age", "BloodPressure", "Glucose"]] = scaler.fit_transform(df[["Age", "BloodPressure", "Glucose"]])
df.head()
print(df)
```

```python
df["RiskScore"] = (df["Age"] + df["BloodPressure"] + df["Glucose"]) / 3
display(df)
```

```python
df.to_csv("cleaned_dataset.cav", index = False)
print("saved")
```
